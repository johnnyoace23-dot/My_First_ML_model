# 📊 Step 2 — Data Visualization

Before training the model, I plotted the relationship between **CGPA** and **IQ**.

### Why?

Visualization helps understand whether the classes are separable.

```python
plt.scatter(df["cgpa"], df["iq"], c=df["placement"])
```

### Observation

- Students with higher CGPA tend to have a higher chance of placement.

# 🎯 Step 3 — Feature Selection

I separated the dataset into **Features (X)** and **Target (Y)**.

```python
X = df.iloc[:,0:2]
Y = df.iloc[:,-1]
```

### Features

- CGPA
- IQ

### Target

- Placement

---

# ✂️ Step 4 — Train-Test Split

I divided the dataset into training and testing data.

```python
x_train, x_test, y_train, y_test = train_test_split(
    X, Y, test_size=0.1
)
```

### Why Train-Test Split?

- **Training Data:** Used to teach the model.
- **Testing Data:** Used to evaluate model performance on unseen data.

**Test Size:** 10%

---

# ⚖️ Step 5 — Feature Scaling

I standardized the features using `StandardScaler`.

```python
scaler = StandardScaler()

x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

### Why Scaling is Important?

CGPA and IQ have different numerical ranges.

Standardization converts values into a common scale:

- Mean = 0
- Standard Deviation = 1

This helps Logistic Regression learn better.

---

# 🤖 Step 6 — Training the Logistic Regression Model

I created my first Machine Learning model.

```python
clf = LogisticRegression()
clf.fit(x_train, y_train)
```

### What Logistic Regression Does?

It predicts the probability that a student belongs to one of two classes:

- **1 → Placed**
- **0 → Not Placed**

It is a **Supervised Classification Algorithm** used for binary classification problems.

---

# 🔮 Step 7 — Model Prediction

After training, I predicted placement on unseen testing data.

```python
y_pred = clf.predict(x_test)
```

### Output

The model returns predicted classes:

- 0
- 1

---

# 📈 Step 8 — Model Evaluation

I measured model performance using Accuracy Score.

```python
accuracy = accuracy_score(y_test, y_pred)
print(accuracy)
```

### Accuracy Formula

Accuracy = Correct Predictions / Total Predictions

A higher score means the model predicts placement correctly for more students.

---

# 🌍 Step 9 — Decision Boundary Visualization

Finally, I visualized how the Logistic Regression model separates both classes.

```python
plot_decision_regions(
    x_train,
    y_train.values,
    clf=clf,
    legend=2
)
```

### What This Plot Shows

- Blue region → One predicted class.
- Red region → Other predicted class.
- The boundary line is where the model changes its prediction.

This helps understand how Logistic Regression makes decisions.

````python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("placement.csv")
# df = df.rename(columns={"Unnamed: 0":"Student_no"})
df = df.iloc[:,1:]
df
````

# Visualisation of understanding the relationship between iq and cgpa for placement

````python
plt.figure(figsize=(10,6))
plt.style.use('ggplot')
plt.scatter(x=df['cgpa'],y=df['iq'],c=df['placement'],cmap="coolwarm")
plt.title("Relation of IQ and CGPA for Getting Placement",weight="bold",color="red")
plt.xlabel("CGPA of Student")
plt.ylabel("IQ of Student")
plt.colorbar(label=df['placement'])
plt.grid(True,alpha=0.4)
````

````python
x = df.iloc[:,0:2]
x
````

````python
y = df.iloc[:,-1]
y
````

# Applying Train and Test split

````python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.1)
````

# Applying Scaling

````python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
x_train = scaler.fit_transform(x_train)
x_test = scaler.fit_transform(x_test)
````

# Model Training

````python
from sklearn.linear_model import LogisticRegression

clf = LogisticRegression()


#training ml model
clf.fit(x_train,y_train)
````

````python
y_pred = clf.predict(x_test)
y_pred
````

````python
y_test
````

# Getting our ML model accuracy score

````python
from sklearn.metrics import accuracy_score

percentage = accuracy_score(y_test,y_pred)

print(f"Our ML model score is : {(percentage*100)} ")
````

# Decision plot of this ML model

````python
from mlxtend.plotting import plot_decision_regions

plot_decision_regions(x_train, y_train.values, clf=clf, legend=2)
````
